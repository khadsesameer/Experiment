**task-2**

```python
visa_df.dtypes   # using for loop
# idea
# extract keys and values
# apply the for loop on values
# for i,j in zip(values,keys)
# if i == 'object':
#     j
```

```python
keys=visa_df.dtypes.keys().tolist()
values=visa_df.dtypes.values.tolist()
cat,num=[],[]
for i,j in zip(keys,values):
    if j=='object':
        cat.append(i)
    else:
        num.append(j)
cat,num
```
